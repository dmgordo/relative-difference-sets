# Relative Difference Sets

Let $G$ be a group of order $mn$, and $N$ a normal subgroup of order $n$. An $(m,n,k,\lambda)$-relative difference set is a subset $R$ of $G$ with $k$ elements such that

$$RR^{(-1)} = k\,1_G + \lambda(G-N)$$

in $\mathbb{Z}[G]$. This repository contains existence results for **cyclic** relative difference sets with $k\leq256$, extending earlier computations of Lam and Pott.

The database `rds.json` uses keys such as `RDS(73,7,64,8)`. A representative is a list of residues modulo $mn$; the forbidden subgroup is $m\mathbb{Z}_{mn}$.

This notebook uses a **standard Python 3 kernel**, without SageMath. `pandas` is optional and used only for prettier tables. Status `Yes` denotes existence, `No` nonexistence, and `Open` an unresolved case.  `All` means that all such sets up to equivalence are known, and included in the database.

## Searching the database

This gives an example of how to examine ranges of results.  Adjust the ranges and statuses below. When pandas is unavailable, `show_tab()` returns ordinary Python row dictionaries.


```python
T = init_tab()
for R in relative_diffsets:
    m = get_m(R)
    n = get_n(R)
    k = get_k(R)
    lam = get_lam(R)
    if get_status(R) in ['All','Yes']: # look at ones we know exist
        if m in range(200):
            if n in range(2,20):
                if k in range(257):
                    if lam in range(1,k):
                        add_tab_entry(T,R)                            
show_tab(T)
```

## Working with and verifying cyclic relative difference sets

`get_rds(m,n,k,lam,i)` returns `[m,n,k,lam,R]`, where `i` is a **zero-based** representative index. `is_rds(D)` checks distinct residues, the necessary parameter equation, and all cyclic difference multiplicities. It requires no SageMath. The example below runs the verification only if the requested representative is present.


```python
D = get_rds(21, 6, 16, 2, 0)
print(D)
if D is not None:
    print('Verified cyclic RDS:', is_rds(D))
```
